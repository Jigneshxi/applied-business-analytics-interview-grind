### 7. Analytical Solutions to Key Numerical Questions

- **Q59 (Dice Sum = 5 or 8):**
  $$\text{Total Outcomes} = 36$$
  $$\text{Sum 5} = \{(1,4), (2,3), (3,2), (4,1)\} \implies 4$$
  $$\text{Sum 8} = \{(2,6), (3,5), (4,4), (5,3), (6,2)\} \implies 5$$
  $$P(\text{Sum 5 or 8}) = \frac{9}{36} = \frac{1}{4} = 0.2500 \quad (25.0\%)\$$

- **Q61 (Jeremy Test Score Z-Score):**
  $$Z = \frac{X - \mu}{\sigma} = \frac{178 - 160}{15} = \frac{18}{15} = 1.20 \quad (88.49^{\text{th}} \text{ percentile})$$

- **Q64 (Supercar Observation Probability in 1 Hour):**
  $$P(\ge 1 \text{ Supercar}) = 1 - P(\text{None in any 20-min}) = 1 - (1 - 0.30)^3 = 1 - 0.343 = 0.6570 \quad (65.70\%)\$$

- **Q57 (Bessel's Correction):**
